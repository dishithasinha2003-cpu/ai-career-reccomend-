```python
import pandas as pd

# Career recommendation data
data = {
    "Python": [1, 1, 0, 1, 0, 1],
    "SQL": [1, 1, 1, 0, 0, 1],
    "MachineLearning": [1, 1, 0, 1, 0, 1],
    "Communication": [0, 1, 1, 1, 1, 0],
    "Career": [
        "Data Scientist",
        "AI Engineer",
        "Business Analyst",
        "Software Developer",
        "HR Executive",
        "Data Analyst"
    ]
}

df = pd.DataFrame(data)

# Separate features and target
X = df.drop("Career", axis=1)
y = df["Career"]

# Train machine learning model
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(X, y)

# Take user input
print("Enter your skills (1 = Yes, 0 = No)")

python = int(input("Do you know Python? "))
sql = int(input("Do you know SQL? "))
ml = int(input("Do you know Machine Learning? "))
communication = int(input("Do you have communication skills? "))

# Make prediction
user_data = [[python, sql, ml, communication]]

prediction = model.predict(user_data)

print("Recommended Career:", prediction[0])
```
