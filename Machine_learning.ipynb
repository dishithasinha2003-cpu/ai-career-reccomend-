````python
#Logistic regression--classification is used when the output we want to predict is a category example:spam or not spam,fraudulent or legimate or customer will stay
### Regression vs Classification

Regression:

- Predicts numerical values.
- Example: 5.2 delivery days.

Classification:

- Predicts categories.
- Example: a status label.


````

````python
#prepare the classification dataset-target is order status
classification_features = [
  "Quantity",
  "DiscountPct",
  "CustomerAge",
  "Category",
  "ProductName",
  "PaymentMethod",
   "City",
  "Country",
  "SalesChannel",
  "DeviceType",
  "Currency",
  "OrderYear",
  "OrderMonth",
  "OrderDay",
  "OrderWeekday"
]

classification_target = "OrderStatus"

if classification_target not in ml_df.columns:
   raise ValueError("OrderStatus target is not available.")

available_classification_features = [
   col for col in classification_features
   if col in ml_df.columns
]

classification_df = ml_df[
  available_classification_features + [classification_target]
].copy()

classification_df = classification_df.dropna(
  subset=[classification_target]
)

X_cls = classification_df[available_classification_features]

y_cls = classification_df[classification_target].astype(str)

print("Classification dataset prepared.")

print("Feature columns:", available_classification_features)

print("X shape:", X_cls.shape)

print("y shape:", y_cls.shape)
````

````python
#check classification target distribuation
status_counts = y_cls.value_counts()

print(status_counts)

plt.figure(figsize=(9, 5))

status_counts.plot(kind="bar")

plt.xlabel("Order Status")
plt.ylabel("Number of Records")
plt.title("Order Status Distribution")

plt.xticks(rotation=45)

plt.show()
````

````python
#split the classifaction dataset
class_counts = y_cls.value_counts()

number_of_classes = y_cls.nunique()

use_stratify = (
   class_counts.min() >= 2
   and int(np.ceil(0.20 * len(y_cls))) >= number_of_classes
   and int(np.ceil(0.80 * len(y_cls))) >= number_of_classes
)

stratify_value = y_cls if use_stratify else None

if not use_stratify:
   print(
       "Warning: Stratified splitting is not suitable "
       "for this class distribution. Using a regular split."
   )

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
   X_cls,
   y_cls,
   test_size=0.20,
   random_state=42,
  stratify=stratify_value
)

print("Training features:", X_train_cls.shape)
print("Testing features:", X_test_cls.shape)

print("\nTraining class distribution:")
print(y_train_cls.value_counts())

print("\nTesting class distribution:")
print(y_test_cls.value_counts())
````

````python
#create classification processsing pipeline
numeric_features_cls = X_train_cls.select_dtypes(
  include=["number"]
).columns.tolist()

categorical_features_cls = X_train_cls.select_dtypes(
  include=["object", "string", "category", "bool"]
).columns.tolist()

numeric_pipeline_cls = Pipeline(
   steps=[
      ("imputer", SimpleImputer(strategy="median")),
      ("scaler", StandardScaler())
   ]
)

categorical_pipeline_cls = Pipeline(
   steps=[
      ("imputer", SimpleImputer(strategy="most_frequent")),
      ("encoder", OneHotEncoder(handle_unknown="ignore"))
   ]
)

preprocessor_cls = ColumnTransformer(
   transformers=[
      ("num", numeric_pipeline_cls, numeric_features_cls),
      ("cat", categorical_pipeline_cls, categorical_features_cls)
   ]
)

print("Classification preprocessing pipeline created.")
````

````python
#train logistic model

logistic_model = Pipeline(
   steps=[
      ("preprocessor", preprocessor_cls),
      ("model", LogisticRegression(max_iter=1000))
   ]
)

logistic_model.fit(
   X_train_cls,
   y_train_cls
)

print("Logistic Regression model trained successfully.")
````

````python
#make predication using logistic
y_pred_logistic = logistic_model.predict(X_test_cls)

print("First 10 predicted statuses:")
print(y_pred_logistic[:10])

print("\nFirst 10 actual statuses:")
print(y_test_cls.iloc[:10].values)
````

````python
#undersand the classification accuracy matrices
#accuracy=correct predication \total predictioon
accuracy = accuracy_score(
   y_test_cls,
   y_pred_logistic
)

print("Logistic Regression Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100)
````

````python
#understand precision Recall:F1-score
Precision answers this question:
Out of all the records the model predicted as a particular class, how many were actually that class?
Recall answers:

Out of all the records that truly belong to a particular class, how many did the model correctly identify?
F1-score combines precision and recall into one metric using their harmonic mean.

These metrics are especially useful when classes are imbalanced."
magine a fraud detection model.

- Precision tells us how many predicted fraud cases were actually fraud.
- Recall tells us how many actual fraud cases the model successfully identified.
- F1-score balances precision and recall.
````

````python
precision = precision_score(
   y_test_cls,
   y_pred_logistic,
  average="weighted",
   zero_division=0
)

recall = recall_score(
   y_test_cls,
   y_pred_logistic,
  average="weighted",
   zero_division=0
)

f1 = f1_score(
   y_test_cls,
   y_pred_logistic,
  average="weighted",
   zero_division=0
)

print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
````

````python
#confusion matrix
cm = confusion_matrix(
   y_test_cls,
   y_pred_logistic,
  labels=logistic_model.classes_
)

plt.figure(figsize=(8, 6))

sns.heatmap(
   cm,
   annot=True,
   fmt="d",
  cmap="Blues",
  xticklabels=logistic_model.classes_,
  yticklabels=logistic_model.classes_
)

plt.xlabel("Predicted Status")
plt.ylabel("Actual Status")
plt.title("Logistic Regression - Confusion Matrix")

plt.show()
````
